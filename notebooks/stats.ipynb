# Article statistics

Inspect downloaded article metadata stored in `data/articles.db`.


```python
from pathlib import Path

import polars as pl
```

```python
DB_PATH = Path("../data/articles.db")  # Adjust if the database lives elsewhere
db_path = DB_PATH.expanduser().resolve()
db_uri = f"sqlite:///{db_path.as_posix()}"
db_uri
```

```python
query = "SELECT * FROM articles"
df = pl.read_database_uri(query, db_uri)
df.head(3)
```

## Articles by journal

```python
df.group_by("journal").len().sort("len", descending=True)
```

## Articles by year

```python
df.group_by("year").len().sort("year")
```

## Articles by source

```python
df.group_by("source").len().sort("len", descending=True)
```

## Abstract coverage

```python
df.group_by("journal").agg(
    pl.col("abstract").is_null().mean().alias("share missing abstract")
).sort("share missing abstract", descending=True)
```

```python
df.group_by("source").agg(
    pl.col("abstract").is_null().mean().alias("share missing abstract")
).sort("share missing abstract", descending=True)
```
